# Compare Sentence Similarities

How to generate vector embeddings using a free open-source model and 
measure how similar two sentences are in meaning using Cosine Distance.

```python
# This example uses sentence-transformer library for sentence similarity
# https://sbert.net/

from sentence_transformers import SentenceTransformer

# 1. Load a pretrained Sentence Transformer model
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

sentence_1 = "The dog chased the ball across the yard."
sentence_2 = "A puppy ran after a toy in the garden."  # Very similar meaning
sentence_3 = "Python is a popular programming language."  # Completely different

# embeddings
v1 = model.encode(sentence_1)
v2 = model.encode(sentence_2)
v3 = model.encode(sentence_3)
similarity_v1_v2 = model.similarity(v1, v2)
similarity_v1_v3 = model.similarity(v1, v3)
similarity_v2_v3 = model.similarity(v2, v3)
print(f"Similarity probability between sentence_1 and sentence_2: {similarity_v1_v2}")
print(f"Similarity probability between sentence_1 and sentence_3: {similarity_v1_v3}")
print(f"Similarity probability between sentence_2 and sentence_3: {similarity_v2_v3}")
```

```python
import os
from dotenv import load_dotenv
from pathlib import Path
from google import genai

load_dotenv()
# 2. Now initialize the client (it will automatically find GEMINI_API_KEY)
client = genai.Client()

text_to_embed = "How do vector databases store embeddings?"
response = client.models.embed_content(model="gemini-embedding-001",contents=text_to_embed)
vector = response.embeddings[0].values
print(f"Text Input:  '{text_to_embed}'")
print(f"Vector Dimensions: {len(vector)}")  # 768
print(f"First 5 numbers:   {vector[:5]}")

```
