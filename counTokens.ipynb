# Count Tokens in a Sentence
LLMs do not read words—they read tokens (sub-word chunks). You will see how different words get split into tokens.

tiktoken is from open ai based on Byte Pair Encoding

cl100k_base: The specific encoding scheme used by models like GPT-3.5-Turbo and GPT-4

```python
!pip install tiktoken
```

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")
print("Encoding:",encoding)
text = "This is an example of LLM tokenization!"
tokens = encoding.encode(text)
for token_id in tokens:
    subword = encoding.decode([token_id])
    print(f"ID {token_id}->{subword}")
```
