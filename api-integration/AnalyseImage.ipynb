# Analyze images and files

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

client = OpenAI()

# Role are of
# system / developer : Sets global instructions, behavior, scope, tone, and system rules.
# user : Provides questions, commands, uploaded files, or data inputs.
# assistant : Contains the model's past generated responses to keep track of chat history.
# tool / function : Returns data execution results (e.g., API calls, SQL results, web searches) back to the model.
response = client.responses.create(
    model="gpt-4o-mini",
    input=[
        {
            "role":"user",
            "content":[
                {
                    "type": "input_text",
                    "text": "Which country flag?",
                },
                {
                    "type": "input_image",
                    "image_url": "https://upload.wikimedia.org/wikipedia/en/thumb/4/41/Flag_of_India.svg/960px-Flag_of_India.svg.png",
                },
            ]
        }
    ]
)

print(response.output_text)
```
