# Prompt Engineering Tree-Of-Thought

Simple example using the classic Game of 24 puzzle: make the number 24 using the numbers 4, 1, 1, 6 with basic arithmetic ($+$, $-$, $\times$, $\div$).


```python
import json
from openai import OpenAI 
from dotenv import load_dotenv

load_dotenv()
client = OpenAI()

prompt="""
Problem : Make 24 using numbers [4,1,1,6]
Give 5 distinct initial operations to try. Make sure you use all given numbers.
Return strictly JSON {"options":["op1","op2","op3","op4","op5"]}
"""
llm_response = client.responses.create(
                model="gpt-5.6",
                input=prompt,
)
print(llm_response.output_text)
```
