## OpenAI SDK 

### Sending request and verifying response



```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()

# The chat completions endpoint is used to send a series of messages 
# representing a conversation to a model, which returns a response.
# Specify which model to use depending on the need. 
# message recives dictionary content role and prompt content
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role":"user","content":"What is OpenAI API?"}]
)

# Refer the image for entire response
print(response)
# To get the actual response contents. 
print(response.choices[0].message.content)

# We can use
# response = client.responses.create(
#    model="gpt-5.6",
#    input="Write a one-sentence bedtime story about a unicorn.",
#)

#print(response.output_text)
```
