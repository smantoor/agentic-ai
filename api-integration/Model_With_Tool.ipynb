# Extend the model with tools
Give the model access to external data and functions by attaching tools. Use built-in tools like web search or file search, or define your own for calling APIs, running code, or integrating with third-party systems.

When you use type:web_search, informs the model that internet is available.The model dynamically decides whether or not to execute a web search based on what the user asks.

If you ask "What is $2 + 2$?", it will answer from knowledge. If you ask "What happened in the news today?", it triggers the web_search tool.

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI()

response = client.responses.create(
    model="gpt-4o-mini",
    tools=[{"type":"web_search"}],  # informs model to use internet
    input="What is the latest news on AI?"
)

print(response.output_text)
```
