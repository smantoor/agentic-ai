# OpenAI SDK Assignment 1 — Configured text-response client
Requirements:
1. Use pydantic-settings to load:
   - OPENAI_API_KEY
   - MODEL_NAME
2. Keep those values in a .env file—never hardcode the key.
3. Create an OpenAI client using your settings.
4. Use client.responses.create(), not parse().
5. Send:
   - a system instruction: “Explain concepts simply for a beginner.”
   - a user prompt: “Explain the difference between Pydantic Field and field_validator in three bullet points.”
6. Print:
   - response.output_text
   - input, output, and total token usage from response.usage
7. Wrap the API call in try/except and print a clear error message if it fails.

```python
from pydantic import BaseModel, Field
from pydantic_settings import BaseSettings
from openai import OpenAI, OpenAIError
from dotenv import load_dotenv

# Load the environment variable
load_dotenv()

class AISettings(BaseSettings):
    openai_api_key: str
    model_name:str = Field(default="gpt-4o-mini")

# Use AISettings to initialise openai client
config = AISettings()

input_text=[
            {
                "role":"system",
                "content":"Explain concepts simply for a beinner"
            },
            {
                "role":"user",
                "content":"Explain the difference between Pydantic Field and field_validator in three bullet points"
            }
        ]

try:
    client = OpenAI(api_key=config.openai_api_key)
    response = client.responses.create(
                    model=config.model_name,
                    input=input_text
                )
    print(response.output_text)
    print(response.usage)
except OpenAIError as oae:
    print(oae)
```
