## Using existing MCP Server without any authentication/authorization
### The public Hugging Face MCP server ([https://huggingface.co/mcp](https://huggingface.co/mcp)) 
Provides AI assistants—such as OpenAI models, Claude, Cursor, and VS Code—with access to the Hugging Face ecosystem. It translates Hugging Face's Hub APIs and hosted AI tools into standard MCP tool calls.

```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client = OpenAI()


response = client.responses.create(
            model='gpt-4o-mini',
            tools = [
                {
                    "type":"mcp",
                    "server_label":"huggingface",
                    "server_description":"Public Hugging Face MCP server",
                    "server_url":"https://huggingface.co/mcp",
                    "require_approval":"never",
                },
            ],
            input="Search for popular text classification models on Hugging Face.",
        )

print(response.output_text)
```
