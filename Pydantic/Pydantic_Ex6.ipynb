## Recursive models

```python
from pydantic import BaseModel
from typing import List, Optional

class FamilyTree(BaseModel):
    name: str
    # recursive model reference
    # Optional means a value can be either a specified type or None
    children: Optional[List["FamilyTree"]] = None

# This informs Pydantic that model to be built after complete class is defined
FamilyTree.model_rebuild()

data = {
    "name":"root",
    "children":[
        {
            "name":"child1",
            "children":[
                {"name":"Grandchild1"},
                {"name":"Grandchild2"},
            ]
        },
        {
            "name":"child2"
        }
    
    ]
}
ftree1 = FamilyTree(**data)
print(ftree1)
```
