## Nested Models

```python
from pydantic import BaseModel, Field, EmailStr

class Address(BaseModel):
    street: str = Field(min_length=5, max_length=50)
    city: str = Field(min_length=2,max_length=20)
    # 6 digits pin code
    # pattern works with string
    postcode: str = Field(pattern=r"^\d{6}$")

class User(BaseModel):
    name: str = Field(min_length=2)
    email:EmailStr
    address:Address


data1 = {
    "name":"Sanjay",
    "email":"smantoor@example.com",
    "address":{
        "street":"AT street",
        "city": "Bengaluru",
        "postcode":"123456"
    }
}

user1 = User(**data1)
print(user1.model_dump())
print(user1.address.city)
```
