## Using Literal, List from Typing library

```python
from pydantic import BaseModel, Field, ValidationError
# typing is a built-in Python library for writing more detailed type hints
from typing import Literal

class Product(BaseModel):
    name: str = Field(min_length=3, max_length=50)
    price: float = Field(gt=0)
    stock: int = Field(ge=0,le=1000)
    # Allowed values with Literal
    category: Literal["electronics","clothing","books"]
    discount_percent:float = Field(default=0,ge=0,le=70)

data1={
    "name":"Sanjay",
    "price":99.9,
    "stock":100,
    "category":"electronics",
    "discount_percent":2.6
}

product1 = Product(**data1)
print(product1.model_dump())

data2={
    "name":"Sanjay",
    "price":-99.9,
    "stock":100,
    "category":"electronics",
    "discount_percent":2.6
}

try:
    product2 = Product(**data2)
except ValidationError as ve:
    print(ve)
```

```python
from pydantic import BaseModel, ValidationError
from typing import List

class Lesson(BaseModel):
    title: str
    duration_minutes: int
    is_free: bool

class Tutorial(BaseModel):
    name: str
    instructor: str
    # List of Lesson objects
    lessons: List[Lesson]

data = {
    "name":"Learn Pydantic",
    "instructor":"xyz",
    "lessons": [
        {"title":"Basic Pydantic","duration_minutes":10,"is_free":True},
        {"title":"Advanced Pydantic","duration_minutes":60,"is_free":False},
    ]
}

tutor1 = Tutorial(**data)
print(tutor1.model_dump())
```
