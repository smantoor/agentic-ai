## Example on built-in types in Pydantic

```python
from pydantic import BaseModel, Field, ValidationError, EmailStr, HttpUrl, PositiveInt

class Contcat(BaseModel):
    email: EmailStr
    website: HttpUrl
    contacts: PositiveInt


contact1 = Contcat(email="smantoor@example.com",website="https://github.com/smantoor",contacts=10)
print(contact1.model_dump())

try:
    contact2 = Contcat(email="smantoorexample",website="github/smantoor",contacts=-10)
except ValueError as ve:
    print(ve)
```
